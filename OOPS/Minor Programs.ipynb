### <b> EVEN / ODD

1. a) Check if number in the list are even/odd using Modulo (%) operator

* Time Complexity - O(1)
* Space Complexity - O(1)

```python
def check_even_odd(numbers):
    return ["Even" if num % 2 == 0 else "Odd" for num in numbers]
num_list = list(map(int, input("Enter numbers separated by space: ").split()))
result = check_even_odd(num_list)
print(num_list)
print("Results:", result)
```

1. b) Check if number in the list are even/odd using Bitwise AND (&) operator

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
def check_even_odd(numbers):
    return ["Even" if num & 1 == 0 else "Odd" for num in numbers]                # Bitwise AND operator, Odd number gives value 1 and Even number gives value 0
num_list = list(map(int, input("Enter numbers separated by space: ").split()))
result = check_even_odd(num_list)
print(num_list)
print("Results:", result)
```

### <b> TABLE OF A NUMBER

2. a) Python Program to print table of a number by ITERATIVE APPROACH

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
def printTable(n):
    for i in range (1, 11): 
        print ("%d * %d = %d" % (n, i, n * i))                     # Multiples from 1 to 10
n = int(input("Enter the number for which table is to be printed:"))
printTable(n)
```

2. b) Python Program to print table of a number by Recursive Approach

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
def printTable(n, i):
    if i == 11:
       return
    print(n, "*", i, "=", n * i)                                    # Calculates and prints n * i and then recursively call for i+1 and then prints until if statement is true
    printTable(n, i + 1)
n = int(input("Enter the number for which table is to be printed:"))
printTable(n, 1)
```

### <b> SUM OF NATURAL NUMBERS

3. a) Python program to find sum of first 'n' natural numbers as per the user input by loop

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
n = int(input("Enter the numbers for which sum of natural numbers is to be printed:"))
sum = 0
for i in range(1, n + 1):   # Calculate the sum of numbers from 1 to n.
    sum += i
print(sum)
```

3. b) Python program to find sum of first 'n' natural numbers as per the user input by RECURSIVE APPROACH

* Time Complexity  - O(n)
* Space Complexity - O(n)

```python
def findSum(n):
    if n == 1:                  # base condition
        return 1
    return n + findSum(n - 1)   # The function calls itself until the base condition is obtained
n = int(input("Enter the numbers for which sum of natural numbers is to be printed:"))
print(findSum(n))
```

3. c) Python program to find sum of first 'n' natural numbers as per the user input by FORMULA BASED APPROACH

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
n = int(input("Enter the numbers for which sum of natural numbers is to be printed:"))
sum = n * (n + 1) // 2
print(sum)
```

* Space Complexity - O(1)

### <b> SWAP TWO NUMBERS

5. a) Python program to swap two numbers using a third variable

* Time Complexity  - O(1)
* Space Complexity - O(1)

```python
a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
temp = a                                    # Swap a and b using temp variable
a = b
b = temp
print(a, b)
```

### <b> CLOSEST TO n and DIVISIBLE BY m

6. a) Python program to find the closest number to n which is divisible by m using ITERATIVE CHECKING

* Time Complexity - O(m)
* Space Complexity - O(1)

```python
def closestNumber(n, m):
    closest = 0
    minDifference = float('inf')
    for i in range(n - abs(m), n + abs(m) + 1):       # Check numbers around n
        if i % m == 0:
            difference = abs(n - i)
            if (difference < minDifference or (difference == minDifference and abs(i) > abs(closest))):
                closest = i
                minDifference = difference
    return closest
n = int(input("Enter the number to be found closest to :"))
m = int(input("Enter the divisior number:"))
print(closestNumber(n, m))
```

6. b) Python program to find the closest number to n which is divisible by m using FINDING QUOTIENT

* Time Complexity - O(1)
* Space Complexity - O(1)

```python
def closestNumber(n, m) :
    q = int(n / m)
    n1 = m * q                    # 1st possible closest number
    if((n * m) > 0) :             # 2nd possible closest number
        n2 = (m * (q + 1))        # Same sign of n and m
    else :
        n2 = (m * (q - 1))        # Different sign of n and m
        
    if (abs(n - n1) < abs(n - n2)) :       # if true, then n1 is the required closest number
        return n1 
    return n2                              # else n2 is the required closest number
n = int(input("Enter the number to be found closest to :"))
m = int(input("Enter the divisior number:"))
print(closestNumber(n, m))
```

### <b> THE DICE PROBLEM

7. a) Python program to find the number on the opposite face of the user given number on a dice using if-else Statement 

* Time Complexity - O(1)
* Space Complexity - O(1)

```python
def oppositeFaceOfDice(n):
    if n == 1:
        return 6
    elif n == 2:
        return 5
    elif n == 3:
        return 4
    elif n == 4:
        return 3
    elif n == 5:
        return 2
    else:
        return 1
n = int(input("Enter the dice number whose opposite is to be found:"))
print(oppositeFaceOfDice(n))
```

* Time Complexity - O(1)
* Space Complexity - O(1)

```python
def oppositeFaceOfDice(n):
    ans = 7 - n
    return ans
n = int(input("Enter the dice number whose opposite is to be found:"))
print(oppositeFaceOfDice(n))
```
