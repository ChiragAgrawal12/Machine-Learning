### Types of Operators in Python
1. Arithmetic Operators: +, -, *, /, %, **, //
2. Comparison Operators: ==, !=, >, <, >=, <=
3. Logical Operators: and, or, not
4. Assignment Operators: =, +=, -=, *=, /=, %=, **=, //=
5. Bitwise Operators: &, |, ^, ~, <<, >>
6. Membership Operators: in, not in
7. Identity Operators: is, is not
8. Ternary Operator: value_if_true if condition else value_if_false

### Types of Operations in Python
1. Unary Operations   : It involves a single operand and performs an operation on it.
2. Binary Operations  : It involves two operands and performs an operation on them.
3. Ternary Operations : It involves three operands and performs an operation based on a condition.

### <b> UNARY OPEARTIONS

```python
a = 10         
b = 5
 
# Unary plus (+)
c = -5
print("Unary Plus:", +c)              # The unary plus doesn't change the value of 'c'

# Unary minus (-)
print("Unary Minus:", -c)             # The unary minus negates the value of 'c'

# Unary logical NOT (!)
d = False
print("Unary Logical NOT:", not d)    # Logical NOT of False is True

# Unary bitwise NOT (~)
e = 1
print("Unary Bitwise NOT:", ~e)       # Bitwise NOT of 1 is -2

# Negation
negated_num = -a
print("Negated number:", negated_num)
```

### <b> BINARY OPERATIONS

```python
# Addition (+)
a = 10
b = 5
c = a + b
print("Addition:", c)
print (f"Addition = {a+b}")

# Subtraction (-)
print(f"Subraction = {a-b}")

# Multiplication (*)
print(f"Multiplication = {a*b}")

# Division (/)
print(f"Division = {a/b}")

# Modulo (%)
print(f"Modulo = {a%b}")

# Exponentiation (**)
print(f"Exponentiation = {a**b}")

# Floor Division (//)
print(f"Floor Division = {a//b}")

# Bitwise AND (&)
print(f"Bitwise AND = {a&b}")

# Bitwise OR (|)
print(f"Bitwise OR = {a|b}")

# Bitwise XOR (^)
print(f"Bitwise XOR = {a^b}")

# Bitwise Left Shift (<<)
print(f"Bitwise Left Shift = {a<<1}")

# Bitwise Right Shift (>>)
print(f"Bitwise Right Shift = {a>>1}")

# Logical AND (and)
print(f"Logical AND =" ,(a > 0) and (b > 0))

# Logical OR (or)
print(f"Logical OR =", (a > 0) or (b > 0))

# Logical NOT (not)
print(f"Logical NOT =", not (a > 0))

# Equal to (==)
print("a == b", a == b)                            # Checks if a is equal to b and returns True or False

# Not equal to (!=)
print("a != b =", a != b)                        # Checks if a is not equal to b and returns True or False

# Greater than (>)
print("a > b =", a > b)                          # Checks if a is greater than b and returns True or False

# Less than (<)
print("a < b =", a < b)                          # Checks if a is less than b and returns True or False

# Greater than or equal to (>=)
print("a >= b =", a >= b)                        # Checks if a is greater than or equal to b and returns True or False

# Less than or equal to (<=)
print("a <= b =", a <= b)                        # Checks if a is less than or equal to b and returns True or False
```

### <b> ASSIGNMENT OPERATIONS

```python
# Assignment Operations
num = 5                          # Assign the value 5 to the variable 'num'
num1 = 10                        # Assign the value 10 to the variable 'num1'

num += 10                        # Increase the value by 10 using the += operator
print("Updated number:", num)

num -= 3                         # Decrease the value by 3 using the -= operator
print("Updated number:", num)

num *= 2                         # Multiply the value by 2 using the *= operator
print("Updated number:", num)

num /= 4                         # Divide the value by 4 using the /= operator
print("Updated number:", num)

num %= 3                         # Get the remainder when divided by 3 using the %= operator
print("Updated number:", num)

num1 **= 2                       # Raise the value to the power of 2 using the **= operator
print("Updated number:", num1)

num1 //= 5                       # Perform floor division by 5 using the //= operator
print("Updated number:", num1)
```

### <b> TERNARY OPERATIONS

#### Ways to use Ternary Operations in Python

1. Nested If-Else Statement
2. Tuple
3. Dictionary
4. Lambda Function
5. Print Statement
6. List Comprehension

```python
# Nested If-Else Statement
num = -5
res = "Positive" if num > 0 else "Negative" if num < 0 else "Zero"
print(res)
```

```python
# Tuple
n = 7
res = ("Odd", "Even")[n % 2 == 0]   # (condition_is_false, condition_is_true)[condition]
print(res)
```

```python
# Dictionary
a = 10
b = 20
m1 = {True: a, False: b}[a > b]   # condition_dict = {True: value_if_true, False: value_if_false}
print(m1)
```

```python
# Lambda Function
a = 10
b = 20
m1 = (lambda x, y: x if x > y else y)(a, b)  # lambda x: value_if_true if condition else value_if_false
print(m1)
```

```python
# Print Statement
a = 10
b = 20
print("a is greater" if a > b else "b is greater")   # print(value_if_true if condition else value_if_false)
```

```python
# List Comprehension
a = [1, 2, 3, 4, 5]
result = ["Even" if x % 2 == 0 else "Odd" for x in a]   # list_comp = [value_if_true if condition else value_if_false for item in iterable]
print(result)
```

### <b> Other Operations

```python
# Membership Operators -- in (Check if a value is present in a sequence) and not in (Check if a value is not present in a sequence)
my_list = [1, 2, 3, 4, 5]
print("List:", my_list)
check_num = int(input("Enter a number to check: "))  # Take input from the user to check if it's in the list
if check_num in my_list:
    print(f"{check_num} is present in the list.")
else:
    print(f"{check_num} is not present in the list.")
```

```python
print("Enter numbers to add to the list using space: ")
new_list = []
new_list = [int(x) for x in input("Enter numbers: ").split()]
print("New List:", new_list)
num = int(input("Check number in the new list: "))
if num in new_list:
    print(f"{num} is present in the list.")
else:
    print(f"{num} is not present in the list.")
```

```python
# Identity Operators -- is (Check if two variables point to the same object) and is not (Check if two variables do not point to the same object)
a = [1, 2, 3]
b = a
c = b
if a is b and a is c:
    print("a, b and c point to the same object.")
else:
    print("a, b and c do not point to the same object.")
    if a is c:
        print("a and c point to the same object.")
    else:
        print("a and c do not point to the same object.")
    if a is b:
        print("a and b point to the same object.")
    else:
        print("a and b do not point to the same object.")
```
